## Import Libraries

```python
import pandas as pd
```

## Import the climatic, agriculture, socioeconomic and Lepto data

```python
# Define the path to the processed files
file_path = f'../../data/Preprocessed_Data/'

leptospirosis_processed = pd.read_csv(f"{file_path}leptospirosis_processed.csv")
climatic_processed = pd.read_csv(f"{file_path}climatic_processed.csv")
agriculture_processed = pd.read_csv(f"{file_path}agriculture_processed.csv")
socioeconomic_processed = pd.read_csv(f"{file_path}socioeconomic_processed.csv")
```

## Consolidate to a single DF

```python
consolidated_dataframe = (leptospirosis_processed.merge(climatic_processed,on=["District","Month","Year"])
                                                 .merge(agriculture_processed,on=["District","Year"])
                                                 .merge(socioeconomic_processed,on=["District","Year"])
                                                )
consolidated_dataframe
```

```python
##saved as a new csv
consolidated_dataframe.to_csv('../../data/Preprocessed_Data/consolidated_dataframe.csv',
                index=False)
```
